```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import category_encoders as ce
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

np.random.seed(42)
plt.style.use('fivethirtyeight')
pd.set_option('display.max_columns', None)
np.set_printoptions(suppress=True, precision=4)

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.DataFrame({
    "color": ["red", "blue", "green", "green", "blue", "red"],
    "city": ["NY", "LA", "NY", "SF", "LA", "NY"],
    "target": [1, 0, 1, 0, 0, 1]
})
display(df)

X = df.drop("target", axis=1)
y = df["target"]
```

#### **Ordinal Encoding**

> Assigns each category an integer

```python
oe_encoder = ce.OrdinalEncoder(cols=["color"])
X_oe_trans = oe_encoder.fit_transform(X, y)
display(X_oe_trans)
```

#### **One-Hot Encoding**

> Creates one binary column per category

```python
ohe_encoder = ce.OneHotEncoder(cols=["color"], use_cat_names=True)
X_ohe_trans = ohe_encoder.fit_transform(X)
display(X_ohe_trans)
```

#### **Binary Encoding**

> Encodes category IDs in binary form.
Good when categories are high-cardinality but still need sparsity.

```python
be_encoder = ce.BinaryEncoder(cols=["city"])
X_be_trans = be_encoder.fit_transform(X)
display(X_be_trans)
```

#### **Base-N Encoding**

> Generalization of BinaryEncoding using any base N.

```python
bne_encoder = ce.BaseNEncoder(cols=["city"], base=3)
X_bne_trans = bne_encoder.fit_transform(X)
display(X_bne_trans)
```

#### **Hashing Encoding**

> Applies a hashing function to map categories into a fixed number of columns.

```python
he_encoder = ce.HashingEncoder(cols=["city"], n_components=8)
X_he_trans = he_encoder.fit_transform(X)
display(X_he_trans)
```

#### **Helmert Encoding**
> Orthogonal contrast coding; compares each category to the mean of subsequent categories.

```python
hme_encoder = ce.HelmertEncoder(cols=["color"])
X_hme_trans = hme_encoder.fit_transform(X, y)
display(X_hme_trans)
```

#### **Sum Encoding**

> Also called deviation encoding. Compares each category to the overall mean.

```python
se_encoder = ce.SumEncoder(cols=["color"])
X_se_trans = se_encoder.fit_transform(X, y)
display(X_se_trans)
```

#### **Polynomial Encoding**

> Produces linear, quadratic, cubic contrasts for ordered categories.

```python
pe_encoder = ce.PolynomialEncoder(cols=["color"])
X_pe_trans = pe_encoder.fit_transform(X, y)
display(X_pe_trans)
```

#### **Backward Difference Encoding**

> Each category is compared to the mean of previous categories.

```python
bde_encoder = ce.BackwardDifferenceEncoder(cols=["color"])
X_bde_trans = bde_encoder.fit_transform(X, y)
display(X_bde_trans)
```

#### **Count Encoding**

> Replaces a category with how often it appears.

```python
ce_encoder = ce.CountEncoder(cols=["city"])
X_ce_trans = ce_encoder.fit_transform(X)
display(X_ce_trans)
```

#### **Target Encoding (Mean Encoding)**

> Replaces each category with the mean target value.

```python
te_encoder = ce.TargetEncoder(cols=["city"])
X_te_trans = te_encoder.fit_transform(X, y)
display(X_te_trans)
```

#### **CatBoost Encoding**

> A target encoder that reduces leakage by encoding each row using only prior rows.
This is the safest target-based encoder.

```python
cbe_encoder = ce.CatBoostEncoder(cols=["city"])
X_cbe_trans = cbe_encoder.fit_transform(X, y)
display(X_cbe_trans)
```

#### **Leave-One-Out Encoding**

> Mean target value for the category minus the row’s own target.
Reduces leakage while still powerful.

```python
looe_encoder = ce.LeaveOneOutEncoder(cols=["city"])
X_looe_trans = looe_encoder.fit_transform(X, y)
display(X_looe_trans)
```

#### **M-Estimate Encoding**

> A Bayesian smoothing approach to target encoding.

```python
mee_encoder = ce.MEstimateEncoder(cols=["city"], m=5)
X_mee_trans = mee_encoder.fit_transform(X, y)
display(X_mee_trans)
```

#### **WOE / Weight of Evidence Encoding (Risk Encoding)**

> Commonly used in credit scoring.

```python
woe_encoder = ce.WOEEncoder(cols=["city"])
X_woe_trans = woe_encoder.fit_transform(X, y)
display(X_woe_trans)
```

#### **James-Stein Encoding**

> Another shrinkage-based target encoder using James-Stein estimation.

```python
jse_encoder = ce.JamesSteinEncoder(cols=["city"])
X_jse_trans = jse_encoder.fit_transform(X, y)
display(X_jse_trans)
```

#### **GLMM Encoding (Generalized Linear Mixed Model)**

> Uses mixed models to encode categories.

```python
glm_encoder = ce.GLMMEncoder(cols=["city"])
X_glm_trans = glm_encoder.fit_transform(X, y)
display(X_glm_trans)
```

#### **Quantile Encoder**

> Encodes categories by quantiles of the target distribution, not just mean.

```python
qe_encoder = ce.QuantileEncoder(cols=["city"], quantile=0.5)
X_qe_trans = qe_encoder.fit_transform(X, y)
display(X_qe_trans)
```

#### **RankHotEncoder**

> Equivalent of One-Hot but columns are sorted by category frequency.
Useful for tree models.

```python
rhe_encoder = ce.RankHotEncoder(cols=["city"])
X_rhe_trans = rhe_encoder.fit_transform(X)
display(X_rhe_trans)
```

#### **Gray Encoder**

> Creates Gray codes for categories (binary encoding with only 1-bit change per step).

```python
ge_encoder = ce.GrayEncoder(cols=["city"])
X_ge_trans = ge_encoder.fit_transform(X)
display(X_ge_trans)
```

#### **Putting It All Together in a Pipeline**

```python
from sklearn.pipeline import Pipeline

pipe = Pipeline([
    ("encoder", ce.CatBoostEncoder(cols=["color", "city"])),
    ("model", LogisticRegression())
])
pipe.fit(X, y)

display(pipe.predict_proba(X))
```
