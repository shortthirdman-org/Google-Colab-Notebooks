```python
# @title Exit Sequence Generation

# Global variable to store exit sequence
exit_sequence = []

def ferris_wheel_game(n, k):
  global exit_sequence  # Reference the global variable
  passengers = list(range(1, n + 1))

  index = 0
  while len(passengers) > 2:
    index = (index + k - 1) % len(passengers)
    exit_sequence.append(passengers.pop(index))

  exit_sequence.extend(passengers)  # Add remaining passengers to the exit sequence

# Parameters: 41 passengers, every 3rd passenger exits
n = 41
k = 3

ferris_wheel_game(n, k)

# Print the exit sequence
print(f"Exit sequence: {exit_sequence}")
```

```python
# @title Frame Sequence Generation

# Total passengers
n = 41

# Initialize the frame sequence
frame_sequence = []

# Initialize the first frame (list from 1 to 41)
frame = list(range(1, n + 1))

# Simulate the frames
exit_index = 0  # Index for tracking the exit sequence

# First frame (initial state)
frame_sequence.append((frame.copy(), "No"))  # Add first frame without highlight

# Generate the frame sequence based on the exit logic
while exit_index < len(exit_sequence):
  # If the first passenger (at cabin 0) is the one exiting, repeat the current frame first
  if frame[0] == exit_sequence[exit_index]:
    # Repeat the frame (frame will be the same as before)
    frame_sequence.append((frame.copy(), "No"))
    # Replace the exited passenger (at the front) with 0
    frame[frame.index(exit_sequence[exit_index])] = 0  # Set the exited passenger to 0
    exit_index += 1  # Move to the next exit in the sequence

  # Rotate the frame left (move the first passenger to the back of the list)
  frame.append(frame.pop(0))

  # Add the current frame to the frame sequence after rotation
  frame_sequence.append((frame.copy(), "No"))

# Now scan the frame sequence and mark consecutive frames that have the same first passenger as highlighted
for i in range(1, len(frame_sequence)):
  # Check if the first passenger is not 0
  if frame_sequence[i][0][0] != 0 and frame_sequence[i][0][0] == frame_sequence[i-1][0][0]:
    # Update the highlight flag for both frames
    frame_sequence[i] = (frame_sequence[i][0], "Yes")
    frame_sequence[i-1] = (frame_sequence[i-1][0], "Yes")

# Now print out the first 10 frames with highlight flags
print("\nFirst 10 frames with highlight flags:")
output_first_frames = [f"Frame {i+1}: {frame_sequence[i][0]} (Highlight: {frame_sequence[i][1]})" for i in range(min(10, len(frame_sequence)))]
for frame in output_first_frames:
  print(frame)

# Now print out the last 10 frames with highlight flags
print("\nLast 10 frames with highlight flags:")
output_last_frames = [f"Frame {len(frame_sequence)-10+i+1}: {frame_sequence[len(frame_sequence)-10+i][0]} (Highlight: {frame_sequence[len(frame_sequence)-10+i][1]})" for i in range(10)]
for frame in output_last_frames:
  print(frame)

# Print the total number of frames generated
print(f"\nTotal frames generated: {len(frame_sequence)}")
```

```python
# @title Exit Event Generation

# Initialize the exit event count list with zeros
exit_events_count = [0] * len(frame_sequence)
exit_events = []

# Print out the length of the input frame_sequence
print(f"Input Length: {len(frame_sequence)}")

# Variable to track the number of exit events
exit_event_counter = 0

# Loop through the frame sequence and check for consecutive highlighted frames
for i in range(1, len(frame_sequence)):
  # Check if two consecutive highlighted frames have the same first passenger
  if frame_sequence[i][1] == "Yes" and frame_sequence[i-1][1] == "Yes":
    # Check if they have the same first passenger
    if frame_sequence[i][0][0] == frame_sequence[i-1][0][0]:
      # Exit event detected at the current frame (second frame of the consecutive pair)
      exit_event_counter += 1
      exit_events.append(i + 1)  # Add 1 to make it 1-based index

  # Store the cumulative count of exit events up to the next frame (i+1)
  if i < len(frame_sequence) - 1:
    exit_events_count[i+1] = exit_event_counter

# Print the detected exit events
if exit_events:
  print("\nDetected Exit Events in Consecutive Highlighted Frames:")
  for event in exit_events:
    print(f"Exit Event detected at Frame {event}")
else:
  print("\nNo exit events detected.")

# Print the total number of exit events detected
print(f"\nTotal number of exit events detected: {len(exit_events)}")

# Print the cumulative exit event counts up to each frame as a data structure (list of tuples)
cumulative_exit_events = [(i + 1, exit_events_count[i]) for i in range(len(exit_events_count))]

# Create an array of cumulative exit event counts
exit_events_list = [exit_events_count[i] for i in range(len(exit_events_count))]

# Print the result as an array
print("Cumulative Exit Event Counts Up To Each Frame:")
print(exit_events_list)

# Function to get exit numbers for a given frame and output as a plain string
def get_exit_numbers_for_frame(frame_number, exit_sequence, exit_events_list):
  # Get the cumulative number of exit events for this frame
  num_exits = exit_events_list[frame_number - 1]  # -1 because list is 0-indexed

  # Extract the exit sequence for the number of exits that have occurred up to this frame
  exit_numbers = exit_sequence[:num_exits]

  # Convert the list of exit numbers to a plain string
  exit_numbers_string = ', '.join(map(str, exit_numbers))

  return exit_numbers_string

# Example: Get exit numbers for the given frame
frame_number = 5
exit_numbers_for_frame = get_exit_numbers_for_frame(frame_number, exit_sequence, exit_events_list)

frame_number = 360
exit_numbers_for_frame = get_exit_numbers_for_frame(frame_number, exit_sequence, exit_events_list)

print(exit_numbers_for_frame)
```

```python
# @title Ferris Wheel Animation

# Import libraries
import numpy as np
import plotly.graph_objects as go

def create_rotating_wheel(frame_sequence, exit_sequence):
  # Colors
  colors = {
    'background': '#f8f9fa',
    'wheel': '#343a40',  # Wheel color
    'spokes': '#6c757d',  # Spokes color
    'box_fill': '#e6f3ff',     # Light blue box fill color
    'box_border': '#80b3ff',    # Darker blue border
    'numbers': '#212529',  # Numbers color
    'highlight': '#0000ff',  # Highlighted color (Blue)
    'title': '#212529',  # Title color
    'exit': 'black',  # Exit text color
    'seq_num': '#0000ff'  # Sequnce number text color (Blue)
  }

  # Set animation speed (milliseconds per frame)
  speed = 400

  # Parameters
  n = 41  # Total numbers
  radius = 7.5  # Radius of the wheel
  center_x, center_y = 0, 0  # Center of the wheel
  square_size = 0.75  # Size of each box

  # Static circle and spokes
  theta = np.linspace(0, 2*np.pi, 100)
  circle_x = center_x + radius * np.cos(theta)
  circle_y = center_y + radius * np.sin(theta)

  # Create the initial figure with static circle and spokes
  fig = go.Figure()

  # Add circle for the wheel
  fig.add_trace(go.Scatter(
    x=circle_x, y=circle_y,
    mode='lines',
    line=dict(color=colors['wheel'], width=3),
    showlegend=False
  ))

  # Draw spokes
  num_supports = 12
  angles_support = np.linspace(0, 2 * np.pi, num_supports, endpoint=False)
  for angle in angles_support:
    x_end = center_x + radius * np.cos(angle)
    y_end = center_y + radius * np.sin(angle)
    fig.add_trace(go.Scatter(
      x=[center_x, x_end], y=[center_y, y_end],
      mode='lines',
      line=dict(color=colors['spokes'], width=1.5, dash='dash'),
      showlegend=False
    ))

  # Angles for the boxes and numbers
  angles = np.linspace(0, 2 * np.pi, n, endpoint=False)
  offset_angle = -np.pi / 2  # Start from top

  # Add static blue boxes and numbers
  for i, number in enumerate(range(1, n + 1)):
    angle = angles[i] + offset_angle
    x = center_x + radius * np.cos(angle)
    y = center_y + radius * np.sin(angle)

    # Draw blue box (static background for the number)
    square_x = [
      x - square_size / 2, x + square_size / 2,
      x + square_size / 2, x - square_size / 2,
      x - square_size / 2
    ]
    square_y = [
      y - square_size / 2, y - square_size / 2,
      y + square_size / 2, y + square_size / 2,
      y - square_size / 2
    ]
    fig.add_trace(go.Scatter(
      x=square_x, y=square_y,
      mode='lines',
      line=dict(color=colors['box_border'], width=1.5),
      fill='toself',
      fillcolor=colors['box_fill'],
      showlegend=False
    ))

    # Add the number only if it's not zero
    if number != 0:
      fig.add_annotation(
        x=x, y=y,
        text=str(number),
        font=dict(size=12, color=colors['numbers'], family='Arial'),
        showarrow=False
      )

  # Create Exit Sign annotation (we'll reuse this in each frame)
  exit_sign = dict(
    x=center_x,
    y=center_y - radius - 1.5,
    text="Exit",
    font=dict(size=16, color=colors['exit'], family='Arial Bold'),
    showarrow=False,
    align='center'
  )

  # Add initial Exit Sign
  fig.add_annotation(exit_sign)

  # Frames for animation: use the provided frame_sequence
  frames = []
  for frame_num, frame_data in enumerate(frame_sequence, start=1):
    # Unpack the tuple into frame (list of passengers) and highlight flag
    frame, highlight = frame_data

    frame_annotations = []

    for i, number in enumerate(frame):
      angle = angles[i] + offset_angle
      x = center_x + radius * np.cos(angle)
      y = center_y + radius * np.sin(angle)

      # Default annotation style
      annotation_style = dict(
        x=x,
        y=y,
        text=str(number) if number != 0 else '',  # Use ternary to check if the number is 0
        font=dict(size=12, color=colors['numbers'], family='Arial'),
        showarrow=False
      )

      # Highlight the first number in cabin 0 if the frame's highlight flag is "Yes"
      if highlight == "Yes" and i == 0 and number != 0:
        annotation_style['font'] = dict(size=14, color=colors['highlight'])  # Highlighted font color
        annotation_style['text'] = f"<b>{number}</b>"  # Bold to make it stand out

      # Add the annotation for the current frame
      frame_annotations.append(annotation_style)

    # Add exit sequence display
    exit_num_list = get_exit_numbers_for_frame(frame_num, exit_sequence, exit_events_list)
    sequence_text = f"{exit_num_list}"
    frame_annotations.append(dict(
      x=center_x,
      y=center_y - radius - 0.8,
      text=f"{sequence_text}",
      font=dict(size=12, color=colors['seq_num'], family='Arial'),
      showarrow=False,
      align='center'
    ))

    # Add Exit Sign to each frame
    frame_annotations.append(exit_sign)

    frames.append(go.Frame(
      data=[],
      layout=go.Layout(
          annotations=frame_annotations
      ),
      name=f'frame_{frame_num}'
    ))

  # Add frames to the figure
  fig.frames = frames

  # Add play/pause buttons for animation (Move buttons down)
  fig.update_layout(
      updatemenus=[{
          'type': 'buttons',
          'direction': 'right',
          'showactive': False,
          'x': 0.48,  # Centered position for Play button
          'y': -0.15,  # Move buttons down
          'xanchor': 'right',
          'yanchor': 'top',
          'buttons': [{
              'label': 'Play',
              'method': 'animate',
              'args': [None, {
                  'frame': {'duration': speed, 'redraw': True},
                  'fromcurrent': True,
                  'mode': 'immediate'
              }]
          }]
      },
      {
          'type': 'buttons',
          'direction': 'right',
          'showactive': False,
          'x': 0.58,  # Centered position for Pause button
          'y': -0.15,  # Move buttons down
          'xanchor': 'right',
          'yanchor': 'top',
          'buttons': [{
              'label': 'Pause',
              'method': 'animate',
              'args': [[None], {
                  'frame': {'duration': 0, 'redraw': False},
                  'mode': 'immediate',
                  'transition': {'duration': 0}
              }]
          }]
      }],

      title=dict(
          text='The Ferris Wheel Problem',
          font=dict(size=24, color=colors['title'], family='Arial Bold'),
          x=0.5,
          y=0.95
      ),
      plot_bgcolor=colors['background'],
      paper_bgcolor=colors['background'],
      width=800,
      height=800,
      xaxis=dict(
          range=[center_x - radius - 2, center_x + radius + 2],
          showgrid=False,
          zeroline=False,
          showticklabels=False
      ),
      yaxis=dict(
          range=[center_y - radius - 1.5, center_y + radius + 2],  # Moved wheel even closer to title
          showgrid=False,
          zeroline=False,
          showticklabels=False,
          scaleanchor='x',
          scaleratio=1
      ),
  )

  return fig

# Start
fig = create_rotating_wheel(frame_sequence, exit_sequence)
fig.show()
```
