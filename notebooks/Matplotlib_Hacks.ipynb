### [Adding Text in Matplotlib and Controlling Its Position](https://medium.com/@shouke.wei/f3a5df33327a)

> Learn how to annotate your plots with text, control its placement, alignment, and styling for clear and informative visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
pd.set_option('display.max_columns', None)

plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = [8, 6]

import warnings
warnings.filterwarnings("ignore")
```

#### **Basic Text Annotation**

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.text(5, 0, "Center of plot", color='red')  # x=5, y=0 in data coordinates
plt.tight_layout()
plt.show()
```

#### **Controlling Alignment**

| Parameter | Options                               | Effect               |
| --------- | ------------------------------------- | -------------------- |
| `ha`      | 'left', 'center', 'right'             | Horizontal alignment |
| `va`      | 'top', 'center', 'bottom', 'baseline' | Vertical alignment   |

```python
plt.plot(x, y)
plt.text(5, 0, "Center", ha='center', va='center', color='blue', fontsize=12)
plt.tight_layout()
plt.show()
```

#### **Adding Text Relative to Axes (Overlay)**

```python
fig, ax = plt.subplots()
ax.plot(x, y)

# Text at top-right corner of subplot
ax.text(0.95, 0.95, "Top-right corner",
        transform=ax.transAxes, ha='right', va='top', color='green', fontsize=12)
plt.tight_layout()
plt.show()
```

#### **Adding Text Above a Line (Dynamic Placement)**

```python
mean_y = np.mean(y)
plt.plot(x, y, label='Sine Wave')
plt.axhline(mean_y, color='orange', linestyle='--', label='Mean')

# Add mean label above the line
plt.text(x[-1]*0.95, mean_y*1.05, f"Mean: {mean_y:.2f}", ha='right', va='bottom', color='orange')
plt.tight_layout()
plt.legend()
plt.show()
```

#### **Multi-Subplot Example**

```python
fig, axes = plt.subplots(2, 1, figsize=(6, 6))
for i, ax in enumerate(axes):
    ax.plot(x, y * (i+1))
    mean_val = np.mean(y * (i+1))

    # Overlay mean annotation top-right
    ax.text(0.95, 0.95, f"Mean: {mean_val:.2f}",
            transform=ax.transAxes, ha='right', va='top', color='tab:blue')
plt.tight_layout()
plt.show()
```
